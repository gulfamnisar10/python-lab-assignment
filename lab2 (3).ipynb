```python
def choose_cheapest(costs):
    cheapest = costs[0]
    for cost in costs:
        if cost < cheapest:
            cheapest = cost
    return cheapest

path_costs = [450, 1200, 300, 600]
result = choose_cheapest(path_costs)
print(result)
```
